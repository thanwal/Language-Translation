```python
import re
import pandas as pd
```

```python
df = pd.read_csv('data/Sentence_pairs_in_English_Hindi.csv')
print(df.head())
```

```python
def clean_english_text(text):
    text = str(text).lower()
    text = re.sub(r'[^a-z\s!.?,]', '', text)
    text = re.sub(r'([!.?,])', r' \1 ', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

def clean_hindi_text(text):
    text = str(text)
    text = re.sub(r'[^\u0900-\u097F\s!.?।]', '', text)
    text = re.sub(r'([!.?।])', r' \1 ', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()
```

```python
df['English'] = df['English'].apply(clean_english_text)
df['Hindi'] = df['Hindi'].apply(clean_hindi_text)

print("Verified Hindi Sample:", df['Hindi'].iloc[0]) 
```

```python
max_threshold = 28 

df = df[
    (df['English'].str.split().str.len() <= max_threshold) & 
    (df['Hindi'].str.split().str.len() <= max_threshold)
]

print(f"Dataset size after filtering outliers: {len(df)}")
```

```python
cleaned_file_path = 'data/Cleaned_Sentence_pairs_Full.csv'
df.to_csv(cleaned_file_path, index=False, encoding='utf-8')

print(f"Cleaned data saved to {cleaned_file_path}")
```
